## Ejercicio 4 / Exercise 4
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`. En este caso debe ser capaz de admitir valores reales, con decimales, para base y altura en español. (PISTA: Recuerda que tendrás que hacer uso del método replace)

*Write a program that calculates the area of a parallelogram (base x height). Also using `input`. In this case, it must be able to accept real values with decimals for base and height in Spanish format (comma as decimal separator). (HINT: Remember that you will need to use the replace method)*

```python
base = input("Enter the base: ")
height = input("Enter the height: ")
base = float(base.replace(",", "."))
height = float(height.replace(",", "."))
area = base * height
print(f"The area of the parallelogram is: {area}")
```
