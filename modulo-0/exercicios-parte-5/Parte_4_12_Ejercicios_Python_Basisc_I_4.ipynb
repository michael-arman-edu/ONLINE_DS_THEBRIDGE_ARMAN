## Ejercicio 2 / Exercise 2
Para el siguiente string se pide imprimir por pantalla los siguientes casos

*For the following string, print the following cases:*

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas / Convert everything to uppercase
2. Pásalo todo a minúsculas / Convert everything to lowercase
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula / Only the first letter of each word capitalized, the rest lowercase
4. Crea una lista dividiéndolo por sus espacios / Create a list by splitting it by spaces
5. Sustituye las comas `,` por puntos y comas `;` / Replace commas `,` with semicolons `;`
6. Elimina las `a` minúsculas / Remove all lowercase `a` characters

```python
var = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
print(var.upper())
print(var.lower())
print(var.title())
print(var.split(' '))
print(var.replace(',',';'))
print(var.replace('a', ''))
```

1. Crea una lista con 3 elementos numéricos / Create a list with 3 numeric elements
2. Añade un cuarto elemento / Add a fourth element
3. Calcula la suma de todos / Calculate the sum of all elements
4. Elimina el segundo elemento de la lista / Remove the second element from the list
5. Añade otro elemento en la posicion 3 de la lista / Add another element at position 3 of the list
6. Crea otra lista con 4 elementos y concátenala a la que ya tenías. / Create another list with 4 elements and concatenate it to the existing one.
7. Ordena la lista de menor a mayor. / Sort the list from smallest to largest.
8. BONUS: Ordena la lista de mayor a menor. / BONUS: Sort the list from largest to smallest.  
(*PISTA: Para hacer 8, tendrás que buscar en la información del método... / HINT: For step 8, you will need to look up the method parameters...*)


NOTA: Pinta por pantalla el resultado de cada paso. / *NOTE: Print the result of each step.*

```python
lista = [1, 2, 3]
print(lista)

lista.append(4)
print(lista)

print(sum(lista))

lista.remove(lista[1])
print(lista)

lista.insert(2, 2.5)
print(lista)

lista2 = [4, 5, 6, 7]
lista3 = lista + lista2
print(lista3)

lista3.sort()
print(lista3)

lista3.sort(reverse=True)
print(lista3)

```
