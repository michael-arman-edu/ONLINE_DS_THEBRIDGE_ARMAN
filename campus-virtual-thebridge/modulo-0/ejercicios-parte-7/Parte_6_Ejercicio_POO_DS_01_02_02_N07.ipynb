* Atributos propios de cada tienda: / *Attributes specific to each store:*

* Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas.
* *(Method to calculate total sales across all months, returning a single number.)*
* Método que calcula la media de ventas de los ultimos meses, por empleado
* *(Method that calculates the average sales of the last months, per employee.)*
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* *(Method that returns a string with the store name and its address.)*
* Método que obtiene las ventas del último mes.
* *(Method that gets the sales of the last month.)*

Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

The method rewrites the sales attribute for the last 3 months and also returns the sales with the increase.)*
