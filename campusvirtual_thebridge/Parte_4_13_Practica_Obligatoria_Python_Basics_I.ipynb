## PARTE I / PART I

En este ejercicio debes simular un chatbot al que le haremos pedidos de pizzas.

*In this exercise you must simulate a chatbot that takes pizza orders.*

1. El chatbot tiene que saludar con un: "Buenas tardes, bienvenido al servicio de pedido online, el precio de nuestra pizzas es de 8,95 para la familiar o de 9,90 para la mediana ¿Cuántas pizzas familiares desea?"  
   *The chatbot must greet with: "Good afternoon, welcome to the online order service, the price of our pizzas is 8.95 for the family size or 9.90 for the medium. How many family pizzas do you want?"*
2. El usuario tiene que introducir un número de pizzas en una variable llamada `pizz_familiar`  
   *The user must enter a number of pizzas in a variable called `pizz_familiar`*
3. El chatbot debe responder: "Estupendo, ¿cuántas pizzas medianas desea?" y guardar el número de pizzas medianas en otra variable `pizz_mediana`  
   *The chatbot must respond: "Great, how many medium pizzas do you want?" and save the number in a variable `pizz_mediana`*
4. Además debe responder el chatbot: "Estupendo, se están preparando 'pizz' pizzas familiares y 'pizz_mediana' pizzas medianas. Dígame su dirección"  
   *The chatbot must also respond: "Great, we are preparing 'pizz' family pizzas and 'pizz_mediana' medium pizzas. Please give me your address"*
5. El usuario tiene que introducir una dirección en formato String en otra variable llamada `direcc`  
   *The user must enter an address as a String in a variable called `direcc`*
6. El programa debe calcular el monto total  
   *The program must calculate the total amount*
7. Respuesta final del chatbot: "Le mandaremos las '(número total de pizzas)' pizzas a la dirección 'direcc'. Serán <precio total a pagar en euros>. Muchas gracias por su pedido."  
   *Final chatbot response: "We will send the '(total number of pizzas)' pizzas to the address 'direcc'. The total will be <total price in euros>. Thank you very much for your order."*

```python
print("Buenas tardes, bienvenido al servicio de pedido online, el precio de nuestra pizzas es de 8,95 para la familiar o de 9,90 para la mediana ¿Cuántas pizzas familiares desea?")
pizz_familiar = int(input())

print("Estupendo, ¿cuántas pizzas medianas desea?")
pizz_mediana = int(input())

print(f"Estupendo, se están preparando {pizz_familiar} pizzas familiares y {pizz_mediana} pizzas medianas. Dígame su dirección")
direcc = input()

total_pizzas = pizz_familiar + pizz_mediana
total_price = (pizz_familiar * 8.95) + (pizz_mediana * 9.90)

print(f"Le mandaremos las {total_pizzas} pizzas a la dirección {direcc}. Serán {total_price:.2f} euros. Muchas gracias por su pedido.")
```

## Parte II / Part II

Crea un programa que realice las siguientes acciones:

*Create a program that does the following:*

1. Solicite al usuario que introduzca 3 números enteros separados por comas (por ejemplo: 1,2,3).  
   *Ask the user to enter 3 integers separated by commas (e.g.: 1,2,3).*
2. Convierta esa entrada en una lista de números enteros.  
   *Convert that input into a list of integers.*
3. Utilizando funciones built-in, determine el número máximo, el número mínimo y la suma total de los números introducidos.  
   *Using built-in functions, determine the maximum number, the minimum number and the total sum of the entered numbers.*
4. Determine la verdad o falsedad de la siguiente afirmación: "El número máximo menos el número mínimo es igual a la suma total dividida por 5".  
   *Determine whether the following statement is true or false: "The maximum number minus the minimum number equals the total sum divided by 5".*
5. Muestre los resultados utilizando print.  
   *Display the results using print.*

```python
entrada = input("Enter 3 integers separated by commas (e.g. 1,2,3): ")
numbers = [int(x) for x in entrada.split(",")]

maximo = max(numbers)
minimo = min(numbers)
suma = sum(numbers)

print("Max:", maximo)
print("Min:", minimo)
print("Sum:", suma)
print("Max - Min == Sum / 5:", (maximo - minimo) == (suma / 5))
```
